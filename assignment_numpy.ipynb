* __Задача 2__: Даны два вектора x и y. Проверить, задают ли они одно и то же мультимножество.  
  Например, для x = np.array([1, 2, 2, 4]), y = np.array([4, 2, 1, 2]) ответ True.

```python
 x = np.array([1, 2, 2, 4])
 y = np.array([4, 2, 1, 2])
```

```python
np.all(np.unique(x, return_counts=True)[1] == np.unique(y, return_counts=True)[1]) and \
np.all(np.unique(x, return_counts=True)[0] == np.unique(y, return_counts=True)[0])
```

```python
 x = [1, 2, 2, 4]
 y = [4, 2, 1, 2]
```

```python
def value_counts(arr):
    hashT = {}
    for elem in arr:
        if elem in hashT:
            hashT[elem] += 1
        else:
            hashT[elem] = 1
    return hashT

value_counts(x) == value_counts(y)
```

```python
X = [[1, 2, 3], [4, 5, 6], [7, 8, 9]]
Y = [[9, 8, 7], [6, 5, 4], [3, 2, 1]]
```

```python
distances = []
start_time = time.time()
for x_row in X:
    row_distances = []
    for y_row in Y:
        distance = (sum((xi - yi) ** 2 for xi, yi in zip(x_row, y_row))) ** 0.5
        row_distances.append(distance)
    distances.append(row_distances)
time.time() - start_time
```
